```python
import pandas as pd
import matplotlib.pyplot as plt # visualizing data
import seaborn as sns
```

```python
df = pd.read_csv('world_population_data.csv')
```

```python
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()  # checking for missing values
```

```python
df.duplicated().sum()  # checking for duplicate values
```

```python
# histplot for Population Distribution

sns.histplot(df["2023 population"], bins= 10, kde=True)
plt.title('Population Distribution')
plt.xlabel('Population')
plt.ylabel('Frequency')
plt.show()
```

```python
# Visualizing total population by continent

df.groupby("continent")["2023 population"].sum().sort_values(ascending=False).plot(kind='bar', color='skyblue')
plt.title('Total Population by Continent in 2023')
plt.xlabel('Continent')
plt.ylabel('Total Population')
plt.xticks(rotation=45)
plt.show()
```

```python
top_10_countries = df.sort_values(by="2022 population", ascending=False).head(10)
```

```python
# Bar plot for Top 10 Most Populated Countries in 2022

top_10_countries.plot(kind='bar', x='country', y='2022 population', color='orange')
plt.title('Top 10 Most Populated Countries in 2022')
plt.xlabel('Country')
plt.ylabel('Population')
plt.xticks(rotation=45)
plt.show()
```

```python

# Top 10 countries and thier growth rate
df.groupby("country")["growth rate"].sum().sort_values(ascending=False).head(10)
```

```python
# Top 5 Largest Countries by Area

top_10_area = df.sort_values(by="area (km²)",ascending=False).head(5)
```

```python
# Line plot for Top 5 Largest Countries by Area

plt.plot(top_10_area['country'], top_10_area['area (km²)'], marker='o', linestyle='-', color='green')
plt.title('country vs area (km²)')
plt.xlabel('country')
plt.ylabel('area (km²)')
plt.xticks
plt.show()
```
