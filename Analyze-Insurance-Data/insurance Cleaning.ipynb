```python
# Sorting the value counts of the sex column in ascending order

sex_Count = df["sex"].value_counts().sort_index(ascending=True)
sex_Count
```

```python
# Visualizing the sex distribution

sex_Count.plot(kind='bar',color =['skyblue', 'lightgreen'])
plt.xlabel("sex")
plt.ylabel("Count")
plt.title("Count of each sex category")
plt.show()
```

```python
# Analyzing the distribution of smokers by sex

smoker_by_sex = df.groupby("sex")["smoker"].count().sort_values(ascending=False)
smoker_by_sex
```

```python
# Visualizing the proportion of smokers by sex with a pie chart

smoker_by_sex.plot(kind="pie", autopct='%1.1f%%', startangle=140, colors=['lightcoral', 'lightskyblue'])
plt.title("Proportion of Smokers by sex")
plt.ylabel("")
plt.show()
```

```python
sns.countplot(x='region', data=df, palette='pastel')
plt.title('Count of each region category')
plt.show()
```

```python
sns.countplot(x='smoker', data=df, palette='pastel')
plt.title('Count of Smokers vs Non-Smokers')
plt.show()
```

```python
# Distribution plot for charges

sns.displot(df['charges'], color='purple', kde=True)
plt.title('Distribution of Insurance Charges')
plt.xlabel('Charges')
plt.ylabel('Density')
plt.show()
```

```python
df.columns
```

```python
# Distribution plot for age

sns.displot(df['age'], color='orange', kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Density')
plt.show()
```

```python
# Distribution plots for all numerical columns

for i in df.columns:
    sns.displot(df[i])
    plt.title('Distribution of '+ " " + i)
    plt.show()
```

```python
colorsforp = ['lightcoral', 'lightskyblue', 'lightgreen', 'orange', 'purple', 'pink']
colorchanger = 0
```

```python
for i in df.columns[:-1]:
    sns.scatterplot(data=df, x=i, y="charges", color=colorsforp[colorchanger % len(colorsforp)])
    plt.title('Scatter plot of '+ " " + i + " vs charges")
    plt.show()
    colorchanger += 1
```
