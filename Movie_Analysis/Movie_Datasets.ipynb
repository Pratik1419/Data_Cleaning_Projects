```python
numeric_cols = ["BudgetUSD","US_BoxOfficeUSD","Opening_Day_SalesUSD","One_Week_SalesUSD","IMDbRating","NumVotesIMDb","NumVotesRT"]
```

```python
for col in numeric_cols:
    sns.histplot(df[col],bins=50, kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
categorical_cols = ["Genre","Country","Director","LeadActor"]
```

```python
for col in categorical_cols:
    df[col].value_counts().head(10).plot(kind='bar', figsize=(10,5))
    plt.title(f'Top 10 {col} Counts')
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Scatter plot: Budget vs US Box Office Sales

sns.scatterplot(data=df, x='BudgetUSD', y='US_BoxOfficeUSD')
plt.title('Budget vs US Box Office Sales')
plt.show()
```

```python
# Scatter plot: IMDb Rating vs US Box Office Sales

sns.scatterplot(data=df, x='IMDbRating', y='US_BoxOfficeUSD', alpha=0.7)
plt.title('IMDb Rating vs US Box Office Sales')
plt.show()
```

```python
# Group by Genre and Release Year to get average box office sales and ratings

genre_year_stats = df.groupby(['Genre', 'ReleaseYear']).agg({
    'US_BoxOfficeUSD': 'mean',
    'IMDbRating': 'mean',
    'NumVotesIMDb': 'sum'
}).reset_index()
```

```python
genre_year_stats
```

```python
# Line plot: Average US Box Office Sales by Genre Over Years

plt.figure(figsize=(14,8))
sns.lineplot(data=genre_year_stats, x='ReleaseYear', y='US_BoxOfficeUSD', hue='Genre')
plt.title('Average US Box Office Sales by Genre Over Years')
plt.show()
```

```python
# Top 10 Actors by Number of Ratings

top_actors = df.groupby("LeadActor")["IMDbRating"].mean().sort_values(ascending=False).head(10)
top_actors
```

```python
# How Many Country by Top 10 Actors have?

top_actors_country = df.groupby("Country")["LeadActor"].count().sort_values(ascending=False).head(10)
top_actors_country
```

```python
# Top 10 Directors by Number of Movies

top_directors = df['Director'].value_counts().head(10).index
top_directors
```
