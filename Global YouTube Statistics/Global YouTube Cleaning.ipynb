```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt # visualizing data
import seaborn as sns
```

```python
df = pd.read_csv("Global YouTube Statistics.csv", encoding='latin-1')
```

```python
category_subscributions = df.groupby('category')['subscribers'].mean().sort_values(ascending=False)
category_subscributions
```

```python
# Visualizing the average subscribers by category

category_subscributions.head(6).plot(kind='bar', figsize=(12,6))
plt.title('Average Subscribers by Category')
plt.xlabel('Category')
plt.ylabel('Average Subscribers')
plt.show()
```

```python
# Analyzing average views by category

channel_views = df.groupby('category')['video views'].sum().sort_values(ascending=False)
channel_views
```

```python
# Visualizing total video views by category

channel_views.head(6).plot(kind='bar', figsize=(12,6))
plt.title('Total Video Views by Category')
plt.xlabel('Category')
plt.ylabel('Total Video Views')
plt.show()
```

```python
# Analyzing channel types By count

channel_type_counts = df["channel_type"].value_counts()
channel_type_counts
```

```python
df.columns
```

```python
Country_counts = df["Country"].value_counts().head(10)
Country_counts
```

```python
# Visualizing distribution of channels by country Of top 10 countries

Country_counts.head(10).plot(kind= "pie", autopct='%1.1f%%', figsize=(8,8))
plt.title('Distribution of Channels by Country')
plt.ylabel('')
plt.show()
```

```python
# Visualizing channel types distribution

channel_type_counts.head(6).plot(kind='pie', autopct='%1.1f%%', figsize=(8,8))
plt.title('Channel Types Distribution')
plt.ylabel('')
plt.show()
```

```python
# Analyzing total subscribers by country

country_subscriptions_total = df.groupby('Country')['subscribers'].sum().sort_values(ascending=False).head(10)
country_subscriptions_total
```

```python
# Visualizing total subscribers by country

country_subscriptions_total.plot(kind='bar', figsize=(12,6))
plt.title('Sum Subscribers by Country')
plt.xlabel('Country')
plt.ylabel('Total Subscribers')
plt.show()
```

```python
# Analyzing average subscribers by created year

df.groupby("created_year")["subscribers"].mean().sort_values(ascending=True)
```

```python
# Analyzing median video views rank by channel type

df.groupby("channel_type")["video_views_rank"].median().sort_values(ascending=True)
```

```python
# Analyzing max subscribers for last 30 days by country

df.groupby("Country")["subscribers_for_last_30_days"].max().sort_values(ascending=False).head(10)
```

```python
# Analyzing total subscribers for last 30 days by created month

df.groupby("created_month")["subscribers_for_last_30_days"].sum().sort_values(ascending=False)
```

```python
# Analyzing standard deviation of video views rank by category

df.groupby("category")["video_views_rank"].std()
```

```python
# Analyzing minimum unemployment rate by country

df.groupby("Country")["Unemployment rate"].min().sort_values(ascending=True).head(15)
```

```python
# Analyzing total uploads by channel type

df.groupby("channel_type")["uploads"].sum().sort_values(ascending=False)
```

```python
# Analyzing locations by Latitude and Longitude

df.groupby(["Latitude", "Longitude"]).size()
```

```python
# Visualizing highest yearly earnings by channel type

sns.barplot(x="channel_type", y="highest_yearly_earnings", data = df, palette="Paired", legend="False")
plt.title('Highest Yearly Earnings by Channel Type')
plt.xlabel('Channel Type')
plt.ylabel('Highest Yearly Earnings')
plt.xticks(rotation=45)
plt.show()
```

```python
# Visualizing subscribers by category and country

df.groupby(["category", "Country"])["subscribers"].sum().unstack().plot(kind="bar", stacked=False)
plt.title('Subscribers by Category and Country')
plt.xlabel('Category')
plt.ylabel('Total Subscribers')
plt.legend(title="Country", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Visualizing Gross tertiary education enrollment vs Unemployment Rate

plt.scatter(df["Gross tertiary education enrollment (%)"], df["Unemployment rate"], alpha=0.5)
plt.xlabel("Gross tertiary education enrollment (%)")
plt.ylabel("Unemployment rate")
plt.title("Scatter Plot of Gross Tertiary Education Enrollment vs Unemployment Rate")
plt.show()
```

```python
# Visualizing Average Video Views Rank by Category Of Top 10

df.groupby("category")["video_views_rank"].sum().sort_values(ascending=False).head(10).plot(kind='bar', figsize=(12,6))
plt.title('Average Video Views Rank by Category')
plt.xlabel('Category')
plt.ylabel('Average Video Views Rank')
plt.show()
```

```python
# Visualizing Average Subscribers by Created Year and Channel Type Of Last 5 Years

df.groupby(["created_year", "channel_type"])["subscribers"].mean().unstack().tail() .plot(kind="bar", stacked=False)
plt.title('Average Subscribers by Created Year and Channel Type')
plt.xlabel('Created Year')
plt.ylabel('Average Subscribers')
plt.xticks(rotation=45)
plt.legend(title="Channel Type", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Visualizing Total Subscribers by Created Year and Month Using Area Chart

df.groupby(["created_year", "created_month"])["subscribers"].sum().unstack().head() .plot(kind="area")
plt.title('Total Subscribers by Created Year and Month')
plt.xlabel('Created Year')
plt.ylabel('Total Subscribers')
plt.legend(title="Created Month", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.show()
```

```python
# Analyzing highest yearly earnings by category

cateogrory_highest_yearly_earnings = df.groupby("category")["highest_yearly_earnings"].max().sort_values(ascending=False).head(10)
cateogrory_highest_yearly_earnings
```

```python
# Visualizing highest yearly earnings by category with a Pie Chart

cateogrory_highest_yearly_earnings.plot(kind='pie', autopct='%1.1f%%', figsize=(8,8))
plt.title('Highest Yearly Earnings by Category')
plt.ylabel('')
plt.show()
```

```python
# Analyzing average subscribers for last 30 days by channel type and created year

channel_created_sub_mean = df.groupby(["channel_type", "created_year"])["subscribers_for_last_30_days"].mean().unstack()
channel_created_sub_mean
```

```python
# Visualizing average subscribers for last 30 days by created month

df.groupby("created_month")["subscribers_for_last_30_days"].mean().plot(kind='line', marker='o')
plt.title('Average Subscribers for Last 30 Days by Created Month')
plt.xlabel('Created Month')
plt.ylabel('Average Subscribers for Last 30 Days')
plt.xticks(rotation=45)
plt.show()
```
