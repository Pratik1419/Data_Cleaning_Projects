```python
import pandas as pd
import matplotlib.pyplot as plt # visualizing data
import seaborn as sns
```

```python
df = pd.read_csv("Netflix Userbase.csv")
```

```python
# Find The Average Revenue By Subscription...

average_revenue_by_subscription = df.groupby("Subscription Type")["Monthly Revenue"].mean().round(2)
print(average_revenue_by_subscription)
```

```python
# find the count of Users From Every Country...

user_count_by_country = df.groupby("Country")["User ID"].count()
print(user_count_by_country)
```

```python
df.columns
```

```python
# Find the Median plan duration acc to age...

median_age_by_plan_duration = df.groupby("Plan Duration")["Age"].median()
print(median_age_by_plan_duration)
```

```python
# Total revenue by Gender...
total_revenue_by_gender = df.groupby("Gender")["Monthly Revenue"].sum()
print(total_revenue_by_gender)
```

```python
# Find the average age by device from the database...

average_age_by_device = df.groupby("Device")["Age"].mean()
print(average_age_by_device)
```

```python
# Show the Age Vs Monthly Revenue (Colored by gender)...

plt.figure(figsize=(10,6))
sns.scatterplot(x="Age", y="Monthly Revenue", hue = "Gender", data=df)
plt.title("Age Vs Monthly Revenue (Colored by gender)")
plt.xlabel("Age")
plt.ylabel("Monthly Revenue")
plt.show()
```

```python
# Give Me The Average Monthly Revenue by Country

average_age_by_country = df.groupby("Country")["Monthly Revenue"].mean().reset_index()
sns.barplot(x="Country", y="Monthly Revenue", data =average_age_by_country)
plt.title("Average Monthly Revenue by Country")
plt.xlabel("Country")
plt.ylabel("Monthly Revenue")
plt.xticks(rotation = 90)
plt.show()
```

```python
# Create a Pariplot of Monthly_Revenue and Age...

numerical_columns = ["Monthly Revenue", "Age"]
sns.pairplot(df[numerical_columns], diag_kind="kde")
plt.suptitle("Pariplot of Monthly Revenue and Age", y=1.02)
plt.show()
```

```python
# Show the Subscription Type Distribution...

sns.countplot(x= "Subscription Type", data= df)
plt.title("Subscription Type Distribution")
plt.xlabel("Subscription Type")
plt.ylabel("Count")
plt.xticks(rotation = 45)
plt.show()
```

```python
# Count the Device Distribution" Using countplot

sns.countplot(x="Device", data=df)
plt.title("Device Distribution")
plt.xlabel("Device")
plt.ylabel("Count")
plt.xticks(rotation = 45)
plt.show()
```
