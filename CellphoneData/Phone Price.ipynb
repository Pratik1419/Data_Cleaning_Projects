```python
# Get the lowest and highest price

Lowestprice = df['Price'].min()
Highestprice = df['Price'].max()

# LowestpriceID = int(df.iloc[150][0])
# print("Lowest Price Phone Details:",(LowestpriceID))

# Print the lowest and highest price

print("Lowest Price Of Phone is:", Lowestprice ,'And', "Highest Price Of Phone is:", Highestprice)
```

```python
# Analyze the relationship between 'Price' and 'Sale' columns

price_sales_mean = df.groupby('Price')['Sale'].mean().sort_values(ascending=False).head(5)
print(price_sales_mean)
```

```python
price_sales_mean.plot(kind='bar', color='skyblue')
plt.title('Top 5 Prices with Highest Average Sales')
plt.xlabel('Price')
plt.ylabel('Average Sales')
plt.show()
```

```python
# Identify the top 5 products with the highest median sales

Sale_productid_median = df.groupby('Product_id')['Sale'].median().sort_values(ascending=False).head(5)
print(Sale_productid_median)
```

```python
sns.boxplot(x='internal mem', y='ram', data=df)
plt.title('Boxplot of RAM by Internal Memory')
plt.xlabel('Internal Memory')
plt.ylabel('RAM')
plt.show()
```

```python
# Visualize the distribution of phone prices

sns.histplot(df['Price'], bins=30, kde=True)
plt.title('Distribution of Phone Prices')
plt.xlabel('Price')
plt.show()
```

```python
# Visualize the relationship between 'ram' and 'Sale' using a histogram

sns.histplot(data = df, x="ram",y="Sale" ,bins=30, kde=True)
plt.title('Distribution of RAM vs Sales Count')
plt.xlabel('RAM')
plt.ylabel('Sales Count')
plt.show()
```
